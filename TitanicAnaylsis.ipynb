# Data Cleaning

```python
titanic_data.head(5)
```

```python
titanic_data.info()
```

```python
# So, here 80% of cabin is null, it is better to drop it than filling
titanic_data.drop(columns=["Cabin"], inplace=True)
```

```python
titanic_data.head()
```

```python
# Fare column missed 1 column, so need to be filled
titanic_data["Fare"] = titanic_data["Fare"].fillna(titanic_data["Fare"].mean(numeric_only=True))
```

```python
# Age column also need to be filled
titanic_data["Age"] = titanic_data["Age"].fillna(titanic_data["Age"].median())
```

```python
titanic_data.info()
```

```python
# Check to so is there any null values

titanic_data.isnull().sum()
```

# Exploratory Data Analysis(EDA)

```python
titanic_data.head()
```

```python
# How many people survived totally vs died totally?
titanic_data["Survived"].value_counts()
```

```python
# Total count of man and woman on the cabin
titanic_data["Sex"].value_counts()
```

```python
# Survival rate by gender
titanic_data.groupby("Sex")["Survived"].mean()
```

```python
# Avg age of survived and non survived people
titanic_data.groupby("Survived")["Age"].mean(numeric_only=True)
```

```python
# Survival rate by passenger class
titanic_data.groupby("Pclass").Survived.mean()
```

```python
# Avg fare paid
avg_fare = titanic_data.Fare.mean(numeric_only=True)
avg_fare
```

```python
# traveling alone or with family
titanic_data["FamilySize"] = titanic_data.SibSp + titanic_data.Parch
titanic_data["Alone"] = titanic_data.FamilySize == 0

# count alone vs with family
titanic_data.Alone.value_counts()
```

```python
# Which embarkation point had the most passengers?
titanic_data.Embarked.value_counts()
```

```python
titanic_data.Embarked.value_counts().idxmax()
```

```python
titanic_data.head()
```

```python
# What is the most common age group on the ship?
titanic_data.Age.value_counts().idxmax()
```

```python
pd.cut(titanic_data['Age'], bins=[0,18,30,50,80], labels=['Child','Young','Adult','Senior']).value_counts()
```

```python
# Top 5 highest fare paying passengers
titanic_data.nlargest(5, "Fare")[['Name', 'Fare', 'Pclass', 'Survived']]
```

```python
# New cleaned file
titanic_data.to_csv("cleaned_titanic.csv", index=False)
```
